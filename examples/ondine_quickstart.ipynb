## 3 · Run the pipeline

Three product descriptions in, a category column out — one builder chain, no boilerplate.

```python
import pandas as pd

from ondine import QuickPipeline

data = pd.DataFrame(
    {
        "product": [
            "Apple iPhone 13 Pro Max 256GB",
            "Samsung Galaxy S22 Ultra 512GB",
            'MacBook Pro 16" M2 32GB RAM',
        ],
        "description": [
            "Latest iPhone with advanced camera system",
            "Premium Android flagship with S Pen",
            "Powerful laptop for professionals",
        ],
    }
)

pipeline = QuickPipeline.create(
    data=data,
    prompt="""Categorize this product into exactly ONE of:
    - Electronics > Smartphones
    - Electronics > Laptops
    - Electronics > Accessories

    Product: {product}
    Description: {description}

    Category:""",
    output_columns=["category"],
    model="llama-3.3-70b-versatile",  # free-tier Groq model; provider auto-detected
)
result = pipeline.execute()
print(result.data[["product", "category"]])
print(
    f"\n{result.metrics.processed_rows} rows processed in {result.metrics.total_duration_seconds:.2f}s"
)
print(f"Total cost: ${result.costs.total_cost:.4f}")
```
